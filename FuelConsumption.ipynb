## 22. Correlation Analysis

Correlation analysis is performed to measure the linear relationship
between important numerical variables.

The correlation coefficient ranges from -1 to +1:

- +1 indicates a strong positive relationship
- 0 indicates little or no linear relationship
- -1 indicates a strong negative relationship

```python
numeric_cols = [
    "ENGINESIZE",
    "CYLINDERS",
    "FUELCONSUMPTION_CITY",
    "FUELCONSUMPTION_HWY",
    "FUELCONSUMPTION_COMB",
    "FUELCONSUMPTION_COMB_MPG",
    "CO2EMISSIONS"
]
df[numeric_cols].corr()
```

### Insight

The correlation analysis helps identify which numerical variables
have stronger linear relationships.

Fuel consumption variables generally show positive relationships
with CO₂ emissions.

Engine size and cylinder count are also associated with fuel
consumption and emissions.

Correlation indicates association, not causation.

## 23. Correlation Heatmap

A heatmap is used to visually represent the correlation between
numerical variables.

```python
plt.figure(figsize=(10,7))

corr = df[numeric_cols].corr()

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```

### Insight

The correlation heatmap provides a visual summary of the relationships
between numerical variables.

The strongest positive relationships can be identified by correlation
values closer to +1.

Fuel consumption variables show strong relationships with CO₂ emissions,
while engine size and cylinder count also show relationships with
fuel consumption.

The heatmap helps identify the variables that may be most useful for
further modeling or prediction.

## 24. Average CO₂ Emissions by Vehicle Class

The average CO₂ emissions are calculated for each vehicle class.

```python
class_co2 = (
    df.groupby("VEHICLECLASS")["CO2EMISSIONS"]
    .mean()
    .sort_values()
)

class_co2
```

```python
plt.figure(figsize=(12,6))

class_co2.plot(kind="bar")

plt.title("Average CO₂ Emissions by Vehicle Class")
plt.xlabel("Vehicle Class")
plt.ylabel("Average CO₂ Emissions")

plt.xticks(rotation=45)

plt.show()
```

### Insight

Average CO₂ emissions vary across manufacturers in this dataset.

These differences should be interpreted carefully because manufacturers
may have different numbers and types of vehicles represented in the
dataset.

Therefore, manufacturer-level differences do not necessarily represent
the performance of the manufacturer's entire vehicle lineup.
